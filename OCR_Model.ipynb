### Text Detection
- **Validation ka Function**

```python
def validate_dynamic_master(image_path, master_path, model):
    try:
        image = Image.open(image_path)
        master=Image.open(master_path)
        
        extracted_text = model.query(image, question="Extract all text from this image")["answer"].strip()
        
        master_text = model.query(master, question="What is the expected master text?")["answer"].strip()
        
        result = {
            "status": "All Good" if extracted_text == master_text else "Reject",
            "query1": extracted_text,
            "query2": master_text,
            "comparison": "Match" if extracted_text == master_text else "Mismatch"
        }
        return result
    
    except Exception as e:
        return {"error": str(e), "status": "Failed"}
```
